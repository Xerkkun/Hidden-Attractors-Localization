# Dynamical-system characterization quickstart

This notebook demonstrates the public, dimension-agnostic trajectory-characterization API on synthetic data. It does not run attractor localization or make a chaos or hiddenness claim.

## Imports

If the package is not installed, run `python -m pip install -e .` from the repository root first.

```python
import numpy as np

from hidden_attractors import compute_trajectory_metrics
```

## Prepare a trajectory

The analysis API accepts time and state arrays separately, so it works with any state dimension.

```python
times = np.linspace(0.0, 20.0, 2000)
states = np.column_stack([
    np.sin(times),
    np.cos(1.7 * times),
    0.25 * np.sin(0.5 * times),
])
```

## Compute finite-time metrics

These summaries characterize the supplied samples; they are not proofs of chaos or hiddenness.

```python
metrics = compute_trajectory_metrics(
    times,
    states,
    t_start=5.0,
    divergence_norm=120.0,
)

{key: metrics[key] for key in [
    "dimension",
    "bounded",
    "range_0",
    "range_1",
    "range_2",
    "psd_entropy_component_0",
]}
```

## Interpret the result

The returned dictionary records finite-time numerical properties of this input only. Supply equilibria explicitly when final-state proximity is relevant.

```python
print(f"samples={metrics['n_samples']} dimension={metrics['dimension']}")
print(f"bounded={metrics['bounded']} diverged={metrics['diverged']}")
```
